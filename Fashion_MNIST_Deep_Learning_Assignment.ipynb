# Deep Learning Assignment — Fashion-MNIST
Complete Fashion-MNIST classification using TensorFlow/Keras.


```python
import tensorflow as tf
import numpy as np
from PIL import Image
from tensorflow import keras
from tensorflow.keras import layers, Sequential
from tensorflow.keras.utils import to_categorical
```

## 1. Download Fashion-MNIST dataset

```python
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.fashion_mnist.load_data()
print('Training images:', train_images.shape)
print('Training labels:', train_labels.shape)
print('Testing images:', test_images.shape)
print('Testing labels:', test_labels.shape)
```

## 2. Check the data

```python
print('Image dimensions:', train_images.ndim)
print('First label:', train_labels[0])
print('First image shape:', train_images[0].shape)
```

```python
import matplotlib.pyplot as plt

plt.imshow(train_images[0], cmap='gray')
plt.axis('off')
plt.show()
```

## 3. Fashion-MNIST class names

```python
class_names = [
    'T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

print(class_names)
print('First image class:', class_names[train_labels[0]])
```

## 4. Encode labels (included for the same workflow as MNIST)

```python
encoded_train_labels = to_categorical(train_labels)
encoded_test_labels = to_categorical(test_labels)

print('Original label:', train_labels[10])
print('Encoded label:', encoded_train_labels[10])
```

## 5. Build the neural network

```python
model = Sequential([
    layers.Dense(512, activation='relu', input_shape=(784,)),
    layers.Dense(10, activation='softmax')
])

model.summary()
```

## 6. Compile the model

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

## 7. Reshape and normalize images

```python
train_images = train_images.reshape((60000, 784)).astype('float32') / 255.0
test_images = test_images.reshape((10000, 784)).astype('float32') / 255.0

print('Training data after preprocessing:', train_images.shape)
print('Testing data after preprocessing:', test_images.shape)
```

## 8. Train the model

```python
history = model.fit(
    train_images,
    train_labels,
    epochs=20,
    batch_size=128,
    validation_split=0.1
)
```

## 9. Evaluate on test data

```python
test_loss, test_accuracy = model.evaluate(test_images, test_labels, verbose=1)
print('Test Loss:', test_loss)
print('Test Accuracy:', test_accuracy)
```

## 10. Plot training accuracy and loss

```python
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

## 11. Predict Fashion-MNIST test images

```python
predictions = model.predict(test_images)
predicted_labels = np.argmax(predictions, axis=1)

for i in range(10):
    print(f'Image {i}: Actual = {class_names[test_labels[i]]}, Predicted = {class_names[predicted_labels[i]]}')
```

## 12. Display one prediction

```python
image_index = 0
plt.imshow(test_images[image_index].reshape(28, 28), cmap='gray')
plt.title(
    f"Actual: {class_names[test_labels[image_index]]} | "
    f"Predicted: {class_names[predicted_labels[image_index]]}"
)
plt.axis('off')
plt.show()
```

## 13. Save the trained model

```python
model.save('fashion_mnist_model.keras')
print('Model saved as fashion_mnist_model.keras')
```

## 14. Predict a custom image (optional)
Use a 28×28-style grayscale fashion image. The image is resized automatically.

```python
def image_classifier(image_path):
    image = Image.open(image_path).convert('L').resize((28, 28))
    image_array = np.array(image).astype('float32') / 255.0
    image_array = image_array.reshape(1, 784)

    prediction = model.predict(image_array, verbose=0)
    predicted_class = np.argmax(prediction[0])
    confidence = float(np.max(prediction[0])) * 100

    print('Predicted:', class_names[predicted_class])
    print(f'Confidence: {confidence:.2f}%')
    return class_names[predicted_class], prediction
```
