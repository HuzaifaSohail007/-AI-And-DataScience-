## Step 1: Import Required Libraries

```python
# Import libraries for data analysis

import pandas as pd
import numpy as np

# Import libraries for Machine Learning

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler

# Import classification models

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

# Import evaluation metrics

from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score
from sklearn.metrics import classification_report, confusion_matrix

# Import SMOTE for balancing the dataset

from imblearn.over_sampling import SMOTE

print("All Libraries Imported Successfully")
```

## Step 2: Load the Dataset

```python
# Load the dataset

df = pd.read_csv("framingham.csv")

# Display first 5 rows

df.head()
```

### Explanation

The dataset is loaded using Pandas.

The `head()` function displays the first 5 rows of the dataset so we can understand the data.

## Step 3: Dataset Analysis

```python
# Check the number of rows and columns

print("Dataset Shape:", df.shape)

# Display column names

print("\nColumn Names:")
print(df.columns.tolist())

# Display information about the dataset

print("\nDataset Information:")
df.info()
```

```python
# Display statistical summary of the dataset

df.describe()
```

### Explanation

- `shape` shows the total number of rows and columns.
- `info()` shows the data types and non-null values.
- `describe()` provides statistical information about the numerical columns.


## Step 4: Check Missing Values

```python
# Check missing values in each column

print(df.isnull().sum())

# Calculate total missing values

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

## Step 5: Handle Missing Values

```python
# Fill missing values using the median

df = df.fillna(df.median())

# Check missing values again

print("Missing Values After Handling:")
print(df.isnull().sum())
```

### Explanation

Some columns contain missing values.

The missing values are filled using the **median** value of each column.

The median is simple to use and is less affected by very high or very low values.

## Step 6: Check Duplicate Values

```python
# Check duplicate rows

print("Total Duplicate Rows:", df.duplicated().sum())
```

## Step 7: Check Whether Encoding is Required

```python
# Check data types

print(df.dtypes)
```

### Explanation

Encoding is required when a dataset contains text or categorical values.

The data types are checked before applying encoding.

For this dataset, the columns are numerical, so additional encoding is not required.

## Step 8: Analyze the Target Variable

```python
# Check the target variable distribution

print(df["TenYearCHD"].value_counts())
```

### Explanation

The target variable is **TenYearCHD**.

- `0` means the negative class.
- `1` means the positive class.

The class distribution is checked to determine whether the dataset is balanced or imbalanced.

## Step 9: Select Features and Target

```python
# Separate features and target variable

X = df.drop("TenYearCHD", axis=1)

y = df["TenYearCHD"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

## Step 10: Split the Dataset

```python
# Split the dataset into training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

### Explanation

The dataset is divided into:

- **80% Training Data**
- **20% Testing Data**

The training data is used to train the models.

The testing data is used to evaluate the final performance.

## Step 11: Feature Scaling

```python
# Create StandardScaler object

scaler = StandardScaler()

# Scale training data

X_train_scaled = scaler.fit_transform(X_train)

# Scale testing data

X_test_scaled = scaler.transform(X_test)

print("Feature Scaling Completed Successfully")
```

### Explanation

Feature Scaling is applied to bring the feature values to a similar scale.

It is useful for models such as:

- Logistic Regression
- KNN
- SVM


## Step 12: Check Dataset Balance

```python
# Check class distribution before SMOTE

print("Class Distribution Before Balancing:")

print(y_train.value_counts())
```

## Step 13: Balance the Dataset Using SMOTE

```python
# Create SMOTE object

smote = SMOTE(random_state=42)

# Balance only the training dataset

X_train_balanced, y_train_balanced = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("Class Distribution After SMOTE:")

print(pd.Series(y_train_balanced).value_counts())
```

### Explanation

The target variable is imbalanced.

SMOTE is used to create additional samples for the minority class.

**Important:** SMOTE is applied only to the training data. The testing data is not changed.

## Step 14: Create Different Classification Models

```python
# Create different models for classification

models = {
    "Logistic Regression": LogisticRegression(),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "SVM": SVC(),
    "Naive Bayes": GaussianNB()
}
```

### Explanation

Different Classification Models are created so that their performance can be compared.

The models used are:

- Logistic Regression
- KNN
- Decision Tree
- Random Forest
- SVM
- Naive Bayes


## Step 15: Train and Compare All Models

```python
# Create an empty list to save results

results = []

# Train each model

for name, model in models.items():

    # Train the model
    model.fit(X_train_balanced, y_train_balanced)

    # Predict test data
    y_pred = model.predict(X_test_scaled)

    # Calculate performance metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    # Save results
    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])

# Create results table

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ]
)

# Sort models by accuracy

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

results_df
```

### Explanation

All models are trained using the same balanced training data.

The models are compared using:

- Accuracy
- Precision
- Recall
- F1-Score


## Step 16: Classification Report and Confusion Matrix

```python
# Generate Classification Report and Confusion Matrix

for name, model in models.items():

    # Predict the test data
    y_pred = model.predict(X_test_scaled)

    print("\nModel:", name)

    # Classification Report
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    # Confusion Matrix
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
```

### Explanation

For every model, the following results are generated:

- Classification Report
- Precision
- Recall
- F1-Score
- Confusion Matrix

The Confusion Matrix shows the number of correct and incorrect predictions.

## Step 17: Select the Best Performing Model

```python
# Select the model with the highest accuracy

best_model_name = results_df.iloc[0]["Model"]

best_accuracy = results_df.iloc[0]["Accuracy"]

print("Best Performing Model:", best_model_name)

print("Best Model Accuracy:", round(best_accuracy, 4))
```

### Explanation

The results table is sorted by Accuracy.

The model at the top of the table is selected as the best-performing model.

## Step 18: Hyperparameter Tuning

```python
# Select the best model for Hyperparameter Tuning

if best_model_name == "Random Forest":

    model_for_tuning = RandomForestClassifier(random_state=42)

    param_grid = {
        "n_estimators": [50, 100, 200],
        "max_depth": [None, 5, 10],
        "min_samples_split": [2, 5]
    }

elif best_model_name == "Decision Tree":

    model_for_tuning = DecisionTreeClassifier(random_state=42)

    param_grid = {
        "max_depth": [None, 3, 5, 10],
        "min_samples_split": [2, 5, 10]
    }

elif best_model_name == "KNN":

    model_for_tuning = KNeighborsClassifier()

    param_grid = {
        "n_neighbors": [3, 5, 7, 9],
        "weights": ["uniform", "distance"]
    }

elif best_model_name == "SVM":

    model_for_tuning = SVC()

    param_grid = {
        "C": [0.1, 1, 10],
        "kernel": ["linear", "rbf"]
    }

elif best_model_name == "Logistic Regression":

    model_for_tuning = LogisticRegression(max_iter=1000)

    param_grid = {
        "C": [0.01, 0.1, 1, 10]
    }

else:

    model_for_tuning = GaussianNB()

    param_grid = {
        "var_smoothing": np.logspace(-11, -7, 5)
    }

# Apply GridSearchCV

grid_search = GridSearchCV(
    estimator=model_for_tuning,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy"
)

# Train the model

grid_search.fit(
    X_train_balanced,
    y_train_balanced
)

# Display best parameters

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross Validation Accuracy:")
print(round(grid_search.best_score_, 4))
```

### Explanation

Hyperparameter Tuning is performed on the best-performing model.

`GridSearchCV` automatically tests different settings and finds the combination that gives the best accuracy.

## Step 19: Evaluate the Final Tuned Model

```python
# Get the best tuned model

tuned_model = grid_search.best_estimator_

# Make predictions

y_pred_tuned = tuned_model.predict(X_test_scaled)

# Calculate final results

tuned_accuracy = accuracy_score(y_test, y_pred_tuned)

tuned_precision = precision_score(
    y_test,
    y_pred_tuned,
    zero_division=0
)

tuned_recall = recall_score(
    y_test,
    y_pred_tuned,
    zero_division=0
)

tuned_f1 = f1_score(
    y_test,
    y_pred_tuned,
    zero_division=0
)

print("FINAL TUNED MODEL RESULTS")

print("Model:", best_model_name)

print("Accuracy:", round(tuned_accuracy, 4))

print("Precision:", round(tuned_precision, 4))

print("Recall:", round(tuned_recall, 4))

print("F1-Score:", round(tuned_f1, 4))
```

## Step 20: Final Classification Report

```python
# Display Classification Report for the final tuned model

print(classification_report(
    y_test,
    y_pred_tuned
))
```

## Step 21: FINAL CONFUSION MATRIX

```python
# Create Confusion Matrix for the final optimized model

cm = confusion_matrix(
    y_test,
    y_pred_tuned
)

print("Final Confusion Matrix:")
print(cm)
```

## Step 22: Compare Before and After Hyperparameter Tuning

```python
# Accuracy before tuning

original_accuracy = results_df.iloc[0]["Accuracy"]

# Create comparison table

comparison = pd.DataFrame({
    "Model Version": [
        "Before Tuning",
        "After Tuning"
    ],
    "Accuracy": [
        original_accuracy,
        tuned_accuracy
    ]
})

comparison
```

### Explanation

This table compares the accuracy of the best model:

- Before Hyperparameter Tuning
- After Hyperparameter Tuning

This helps us check whether the tuning process improved the model performance.
